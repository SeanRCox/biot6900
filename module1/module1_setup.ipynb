### C1 — PubMed (via `Bio.Entrez`)

```python
from Bio import Entrez
Entrez.email = "cox.sea@northeastern.edu"
handle = Entrez.esearch(db="pubmed", term="Bipolar genome", retmax=5)
record = Entrez.read(handle)
handle.close()
print("PubMed IDs:", record["IdList"])
```

**TODO (C1) — Note:** which search term did you use, and how many PMIDs came back?

_Your note:_ I used the term 'Bipolar genome' and got 5 results. This is because the handle was created with the 'retmax = 5' option.

### C2 — UniProt (REST API)

```python
import requests
acc = "P04637"   # human tumor-suppressor p53;
r = requests.get(f"https://rest.uniprot.org/uniprotkb/{acc}.json")
data = r.json()
name = data["proteinDescription"]["recommendedName"]["fullName"]["value"]
print("Accession:", acc)
print("Protein:  ", name)
```

**TODO (C2) — Note:** which protein did you retrieve?

_Your note:_  p53, which is perhaps the most critical tumor-suprresing proteins in the human body.

### C3 — PDB (RCSB REST API)

```python
import requests
pdb_id = "1TUP"   # a p53 structure;
r = requests.get(f"https://data.rcsb.org/rest/v1/core/entry/{pdb_id}")
print("PDB ID:  ", pdb_id)
print("Title:   ", r.json()["struct"]["title"])
print("Coordinates:", f"https://files.rcsb.org/download/{pdb_id}.pdb")
```

**TODO (C3) — Note:** what structure did you retrieve?

_Your note:_ the p53 DNA-binding domain bound to a DNA double helix. This is the data of the physical protein structure.

### C4 — GWAS Catalog (EBI REST API)

```python
import requests
base = "https://www.ebi.ac.uk/gwas/rest/api"
rsid = "rs334"  
r = requests.get(f"{base}/singleNucleotidePolymorphisms/{rsid}")
data = r.json()
print("Variant:        ", data.get("rsId"))
print("Functional class:", data.get("functionalClass"))
```

**TODO (C4) — Note:** which variant did you look up, and one thing you noticed?

_Your note:_ I looked us rs334. I found this one interesting because it is a single amino acid change that causes sickle cell disease, which offers malaria resistance at a heavy risk factor.
